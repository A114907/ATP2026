```python
#aluno = (nome, id, [notaTPC, notaProj, notaTeste])
#turma = [aluno]

def criarTurma(Turma):
    N = int(input("Quantos alunos tem a Turma?"))
    while N > 0:
        notas = []
        nome = input("Introduza o nome do aluno")
        id = int(input("Introduza o id do aluno"))
        notaTPC = float(input("Introduza a nota do TPC"))
        notas.append(notaTPC)
        notaProj = float(input("Introduza a nota do projeto"))
        notas.append(notaProj)
        notaTeste = float(input("Introduza a nota do Teste"))
        notas.append(notaTeste)
        aluno = (nome, id, notas)
        Turma.append(aluno)
        N = N - 1
    return Turma


def ConsultaID(Turma):
    id = int(input("Qual é o ID do aluno que deseja consultar?"))
    encontrado = False
    for aluno in Turma:
        if aluno[1] == id and not encontrado:
            print(f"Nome:  {aluno[0]}  ;  ID:  {aluno[1]}  ;  Nota TPC:  {aluno[2][0]}  ;  Nota Projeto:  {aluno[2][1]}  ;  Nota Teste:  {aluno[2][2]}")
            encontrado = True
    if encontrado == False:
        print("Esse Aluno não se encontra na Turma.")

def GuardarTurma(Turma, fnome):
    file = open(fnome, "w")
    for aluno in Turma:
        nome, id, [notaTPC, notaProj, notaTeste] = aluno
        file.write(f"Nome: {nome} :: ID: {id} :: Nota TPC: {notaTPC} :: Nota Projeto: {notaProj} :: Nota Teste: {notaTeste}")
        file.write("\n")
    file.close()
  

def Carregaficheiro(fnome):
    Turma = []
    file = open(fnome, "r")
    for line in file:
        campos = line.split(" :: ")
        notas = []
        nome = campos[0].split(": ")[1]
        id = int(campos[1].split(":")[1])
        notaTPC = float(campos[2].split(": ")[1])
        notaProj = float(campos[3].split(": ")[1])
        notaTeste = float(campos[4].split(": ")[1])
        notas = [notaTPC, notaProj, notaTeste]
        aluno = (nome, id, notas)
        Turma.append(aluno)
    file.close()
    return Turma



def menu():
    print("""
    1: Criar uma turma;
    2: Inserir um aluno na turma;
    3: Listar a turma;
    4: Consultar um aluno por id;
    8: Guardar a turma em ficheiro;
    9: Carregar uma turma dum ficheiro;
    0: Sair da aplicação
    """)


Turma = []
menu()
n = int(input("Qual operação deseja utilizar?"))
while n != 0:
    if n == 1:
        Turma = criarTurma(Turma)
        print(Turma)
    elif n == 2:
        notas = []
        nome = input("Introduza o nome do aluno")
        id = int(input("Introduza o id do aluno"))
        notaTPC = float(input("Introduza a nota do TPC"))
        notas.append(notaTPC)
        notaProj = float(input("Introduza a nota do projeto"))
        notas.append(notaProj)
        notaTeste = float(input("Introduza a nota do Teste"))
        notas.append(notaTeste)
        aluno = (nome, id, notas)
        Turma.append(aluno)
        print(Turma)
    elif n == 3:
        for aluno in Turma:
            print(f"Nome:  {aluno[0]}  ;  ID:  {aluno[1]}  ;  Nota TPC:  {aluno[2][0]}  ;  Nota Projeto:  {aluno[2][1]}  ;  Nota Teste:  {aluno[2][2]}")
    elif n == 4:
        ConsultaID(Turma)
    elif n == 8:
        GuardarTurma(Turma, "turma.txt")
    elif n == 9:
        print(Carregaficheiro("turma.txt"))
    else:
        print("Erro! Operação inválida")
    menu()
    n = int(input("Qual operação deseja utilizar?"))
print(f"A tua aplicação terminou e de momento a turma é {Turma}")
    
    
        
        

```
