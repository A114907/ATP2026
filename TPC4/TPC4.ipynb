```python
def criaLista(N):
    lista = []
    i = 0
    while i < N:
        import random
        n = random.randint(1,100)
        lista.append(n)
        i = i + 1
    return lista


def leLista(K):
    lista = []
    for i in range(1,K+1):
        num = int(input("Introduza um número: "))
        lista.append(num)
    return lista


def somaLista(lista):
    soma = 0
    for e in lista:
        soma = soma + e
    return soma


def mediaLista(lista):
    soma = 0
    i = 0
    if len(lista) == 0:
        media = 0
    else:
        for e in lista: 
            soma = soma + e
            i = i + 1
        media = soma / i 
    return media


def maiorLista(lista):
    maior = lista [0]
    for e in lista:
        if e > maior:
            maior = e
    return maior


def menorLista(lista):
    menor = lista [0]
    for e in lista:
        if e < menor:
            menor = e
    return menor


def estaOrdenadaC(lista):
    i = 0
    res = "Sim"
    while i < (len(lista) - 1) and res:
        if lista[i] > lista[i + 1]:
            res = "Não"
        i = i + 1
    return res


def estaOrdenadaD(lista):
    i = 0
    res = "Sim"
    while i < (len(lista)-1) and res:
        if lista[i] < lista [i + 1]:
            res = "Não"
        i = i + 1
    return res


def indiceDe(lista, elem):
    encontrei = False
    i = 0
    res = - 1
    while i < len(lista) and not encontrei:
        if lista[i] == elem:
            encontrei = True
            res = i
        i = i + 1
    return res



lista = []
n = int(input("""Qual número deseja primir?
(1) Criar Lista
(2) Ler Lista
(3) Soma
(4) Média
(5) Maior
(6) Menor
(7) estaOrdenada por ordem crescente
(8) esta Ordenada por ordem decrescente
(9) Procura um elemento
(0) Sair"""))
while n != 0:
    if n == 1:
        N = int(input("Por quantos números deseja que a sua lista seja composta?"))
        lista = criaLista(N)
        print(lista)
    elif n == 2:
        K = int(input("Por quantos números deseja que a sua lista seja composta?"))
        lista = leLista(K)
        print(lista)
    elif n == 3:
        print(somaLista(lista))
    elif n == 4:
        print(mediaLista(lista))
    elif n == 5:
        print(maiorLista(lista))
    elif n == 6:
        print(menorLista(lista))
    elif n == 7:
        print(estaOrdenadaC(lista))
    elif n == 8:
        print(estaOrdenadaD(lista))
    elif n == 9:
        elem = int(input("Qual elemento deseja procurar?"))
        print(indiceDe(lista, elem))
    else:
        print("Erro! Número inválido.")
    n = int(input("""Qual número deseja primir? 
    (1) Criar Lista
    (2) Ler Lista
    (3) Soma
    (4) Média
    (5) Maior
    (6) Menor
    (7) estaOrdenada por ordem crescente
    (8) esta Ordenada por ordem decrescente
    (9) Procura um elemento
    (0) Sair """))
print(f"A aplicação terminou e a sua lista é {lista}")
```
