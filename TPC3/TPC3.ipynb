```python
def soma(N,a):
    total = a
    while total < N:
        n = int(input("Introduza um número (1 a 10):"))
        print(n)
        total = total + n
        print(f"O Total está em {total}")
        if total == N:
            print(f"Ganhaste! Atingiste primeiro os {N}!")
        else:
            subtração = 11 - n
            print(f"O computador jogou {subtração}")
            total = total + subtração
            print(f"O Total está em {total}")
    return N

print("Vamos jogar ao Corrida para o 100")
res = input("Prima 1 se quer que o computador comece primeiro ou prima 2 se quer começar primeiro")
if res == "1":
    print("O computador jogou 1")
    soma(100,1)
    print("Ganhei! Atingi primeiro os 100!")
elif res == "2":
    total = 0
    while total < 100:
        n = int(input("Introduza um número (1 a 10):"))
        print(n)
        total = total + n
        print(f"O Total está em {total}")
        if total == 100:
            print("Ganhaste! Atingiste primeiro os 100!")
        else:
            resto = (total - 1) % 11
            if resto == 0:
                num = 1
            else:
                num = 11 - resto 
            print(f"O computador jogou {num}")
            total = total + num
            print(f"O Total está em {total}")
            if total == 100:
                print("Ganhei! Atingi primeiro os 100!")
else:
    print("Erro, apenas pode primir 1 e 2!")

    
```
